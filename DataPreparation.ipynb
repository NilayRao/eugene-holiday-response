```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

```python
init = pd.read_csv('call_data_from_CAD.csv')
init
```

```python
import pandas as pd

# List of years to load
years = range(2014, 2026)

# Folder where the files are stored
folder = "time/"

# Load and combine all files
new2 = []
for year in years:
    path = f"{folder}class_data_{year}.csv"
    df_year = pd.read_csv(path, low_memory=False)
    df_year['year'] = year  # optional: add a year column
    new2.append(df_year)

# Concatenate all dataframes
classdata = pd.concat(new2, ignore_index=True)
classdata
```

```python
# Make sure both IDs are strings for a reliable merge
init['IncidentNumber'] = init['IncidentNumber'].astype(str)
classdata['inci_id'] = classdata['inci_id'].astype(str)

# Merge on IncidentNumber (init) and inci_id (classdata)
df = pd.merge(classdata, init, left_on='inci_id', right_on='IncidentNumber', how='left')
df = df[df['secs_to_arrv'] >= 0]
df
```

```python
import pandas as pd
import holidays

# Ensure datetime format
df['calltime'] = pd.to_datetime(df['calltime'])

# Create a 'date_only' column with just the date
df['date_only'] = df['calltime'].dt.date

# Rebuild the holidays map with proper years
us_holidays = holidays.US(years=range(df['date_only'].min().year, df['date_only'].max().year + 1))

# Map holidays to the new 'date_only' column
df['holiday_name'] = df['date_only'].map(us_holidays)
df['is_holiday'] = df['holiday_name'].notna().astype(int)
df
```

```python
df_cahoots = df[df['primeunit'] == 'CAHOOT'].reset_index(drop=True)
df_cahoots
```

```python
not_df = df[df['primeunit'] != 'CAHOOT'].reset_index(drop=True)
not_df
```

# Export dataframes to csv's to be used in Data Analysis Notebook

```python
df.to_csv("df.csv", index=False)
df_cahoots.to_csv("cahoots.csv", index=False)
not_df.to_csv("other.csv", index=False)
```
